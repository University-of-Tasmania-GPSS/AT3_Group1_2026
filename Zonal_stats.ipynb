```python
# Calculating roof area with a slope less than 65 degrees using the raster calculator in QGIS.

from sklearn import preprocessing
from qgis import processing
from qgis.core import QgsCoordinateReferenceSystem


layers = [
    (
        "C:/Users/jesta/OneDrive - University of Tasmania/Data analytics/"
        "Assignments/AT3/SEBE results_east/Energyyearroof.tif"
    ),
    (
        "C:/Users/jesta/OneDrive - University of Tasmania/Data analytics/"
        "Assignments/AT3/Zonal_Stats/East_Wall_slope.tif"
    ),
]

expression = (
    '"0.5 * "EnergyyearroofE@1" / cos((("East_Wall_slope@1" < 65) * "East_Wall_slope@1") * 3.141 / 180)'
)

processing.run(
    "native:rastercalc",
    {
        "LAYERS": layers,
        "EXPRESSION": expression,
        "EXTENT": (
            "534002.752959464,534603.252959464,"
            "5249503.494062109,5250103.994062109 [EPSG:7855]"
        ),
        "CELL_SIZE": 0.5,
        "CRS": QgsCoordinateReferenceSystem("EPSG:7855"),
        "OUTPUT": (
            "C:/Users/jesta/OneDrive - University of Tasmania/Data analytics/"
            "Assignments/AT3/Zonal_Stats/East_roof_area3.tif"
        ),
    },
)
```
